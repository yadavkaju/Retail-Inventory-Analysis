```python
import pandas as pd
import os 
from sqlalchemy import create_engine
import sqlite3
```

```python
engine = create_engine(
    "mysql+pymysql://root:kajal@localhost:3306/vendor_db"
)
```

```python
def ingest_db_chunked(file_path, table_name, engine):
    for chunk in pd.read_csv(file_path, chunksize=5000):
        chunk.columns = chunk.columns.str.lower().str.replace(' ', '_')
        
        chunk.to_sql(
            table_name,
            engine,
            if_exists='append',
            index=False
        )

for file in os.listdir('data'):
    if file.endswith('.csv'):
        path = os.path.join('data', file)
        table_name = file.replace('.csv', '')
        
        ingest_db_chunked(path, table_name, engine
```

```python
#Aggregate SALES
sales_summary = pd.read_sql("""
    SELECT
        inventoryid,
        store,
        SUM(salesquantity) AS total_sales_qty,
        SUM(salesdollars) AS total_sales_amount,
        SUM(excisetax) AS total_excise_tax
    FROM sales
    GROUP BY inventoryid, store
""", engine)

```

```python
#PURCHASE AGGREGATION

purchase_summary = pd.read_sql("""
    SELECT
        inventoryid,
        store,
        SUM(quantity) AS total_purchase_qty,
        SUM(dollars) AS total_purchase_amount,
        AVG(purchaseprice) AS avg_purchase_price
    FROM purchases
    GROUP BY inventoryid, store;

""",engine)
```

```python
begin_inventory_summary = pd.read_sql("""
    SELECT
        inventoryid,
        store,
        onhand AS begin_qty
    FROM begin_inventory;

""",engine)
```

```python
end_inventory_summary = pd.read_sql("""
    select 
        inventoryid,
        store,
        onhand as end_qty
    from end_inventory
""",engine)
```

```python
final_sales_summary = (
    begin_inventory_summary
    .merge(purchase_summary, on=['inventoryid','store'], how='left')
    .merge(sales_summary, on=['inventoryid','store'], how='left')
    .merge(end_inventory_summary, on=['inventoryid','store'], how='left')
)


```

```python
final_sales_summary.fillna(0, inplace=True)
```

```python
final_sales_summary['calculated_end_qty'] = (
    final_sales_summary['begin_qty']
    + final_sales_summary['total_purchase_qty']
    - final_sales_summary['total_sales_qty']
)
```

```python
final_sales_summary['shrinkage'] = (
    final_sales_summary['end_qty']
    - final_sales_summary['calculated_end_qty']
)
```

```python
final_sales_summary['gross_profit'] = (
    final_sales_summary['total_sales_amount']
    - final_sales_summary['total_purchase_amount']
)
```

```python
final_sales_summary['profit_margin_pct'] = (
    final_sales_summary['gross_profit'] /
    final_sales_summary['total_sales_amount'].replace(0, pd.NA)
) * 100
```
